## Setup REINVENT score (from REINVENT4)

```python
from rdkit import Chem
import numpy as np

from reinvent.scoring.scorer import Scorer
from reinvent.utils import config_parse

scoring_config = """
[scoring]
type = "geometric_mean"

[[scoring.component]]
[scoring.component.custom_alerts]

[[scoring.component.custom_alerts.endpoint]]
name = "Alerts"

params.smarts = [
    "[*;r8]",
    "[*;r9]",
    "[*;r10]",
    "[*;r11]",
    "[*;r12]",
    "[*;r13]",
    "[*;r14]",
    "[*;r15]",
    "[*;r16]",
    "[*;r17]",
    "[#8][#8]",
    "[#6;+]",
    "[#16][#16]",
    "[#7;!n][S;!$(S(=O)=O)]",
    "[#7;!n][#7;!n]",
    "C#C",
    "C(=[O,S])[O,S]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#16;!s]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#7;!n]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#8;!o]",
    "[#8;!o][C;!$(C(=[O,N])[N,O])][#16;!s]",
    "[#8;!o][C;!$(C(=[O,N])[N,O])][#8;!o]",
    "[#16;!s][C;!$(C(=[O,N])[N,O])][#16;!s]"
]

[[scoring.component]]
[scoring.component.QED]

[[scoring.component.QED.endpoint]]
name = "QED"
weight = 0.6


[[scoring.component]]
[scoring.component.NumAtomStereoCenters]

[[scoring.component.NumAtomStereoCenters.endpoint]]
name = "Stereo"
weight = 0.4

transform.type = "left_step"
transform.low = 0
"""

toml_config_filename = "reinvent_scoring.toml"

with open(toml_config_filename, "w") as tf:
    tf.write(scoring_config)

config = config_parse.read_config(toml_config_filename, "toml")

reinvent_scoring_function = Scorer(config['scoring'])

def is_valid_mol(mol: Chem.Mol | None) -> float:
    if mol is None:
        return 0.0
    try:
        test_mol = Chem.Mol(mol)
        Chem.SanitizeMol(test_mol)
        return 1.0
    except Exception:
        return 0.0


def scoring_function(mols: list[Chem.Mol | None])->list[float]:
  smilies = []
  invalid_mask = []
  duplicate_mask = []

  for mol in mols:
    valid_flag = is_valid_mol(mol)
    if valid_flag == 1.0:
      invalid_mask.append(1)
      smiles = Chem.MolToSmiles(mol)
      if smiles in smilies:
        duplicate_mask.append(0)
      else:
        duplicate_mask.append(1)
      smilies.append(smiles)
    else:
      smilies.append("")
      invalid_mask.append(0)
      duplicate_mask.append(0)

  invalid_mask = np.array(invalid_mask)
  duplicate_mask = np.array(duplicate_mask)

  score_results = reinvent_scoring_function(smilies, invalid_mask, duplicate_mask)
  scores = score_results.total_scores

  return scores.tolist()
```

## Tuning

```python
import torch
from rdkit import Chem, RDLogger
from ml_conformer_generator.src.mlconfgen import MLConformerGenerator, evaluate_samples
from ml_conformer_generator.src.mlconfgen.utils import standardize_mol
# from ml_conformer_generator.pharmacophore import color_tanimoto

RDLogger.DisableLog('rdApp.*')

if torch.cuda.is_available():
    device = torch.device("cuda:0")
elif torch.backends.mps.is_available():
    device = torch.device("mps:0")
else:
    device = torch.device("cpu")

print(f"Intitialising model on {device}")

ref_mols = [x for x in Chem.SDMolSupplier('./rl_fine_tuning_refs.sdf')]
# diffusion_steps = [10, 20, 50, 100]

diffusion_steps = [100]
for ref_mol in ref_mols[1:]:
  for d_steps in diffusion_steps:
    ref_name = ref_mol.GetProp('_Name')
    print(f"Starting Fine Tuning for {ref_name} at {d_steps} diffusion steps")

    generator = MLConformerGenerator(
                                    edm_weights="./edm_moi_chembl_15_39.pt",
                                    adj_mat_seer_weights="./adj_mat_seer_chembl_15_39.pt",
                                    device=device,
                                    diffusion_steps=d_steps,
                                    )

    # ref_mol = Chem.MolFromMolFile('./ml_conformer_generator/assets/demo_files/ceyyag.mol')


    generator.fine_tune(
                      scoring_function=scoring_function,  # This should output normalised score from (0, 1)
                      reference_conformer=ref_mol,
                      variance=1,
                      # RL Fine-tune params
                      n_epochs=20,
                      train_batch_size=256,
                      eval_batch_size=256,
                      learning_rate= 5e-5,
                      sigma=128.0,
                      lambda_edm_adapter=1,
                      lambda_edm_reg=0.02,
                      temperature=1.1,
                      n_samples_per_mol=32,
                      eval_every=2,
                      save_dir=f"./{ref_name}_d_steps_{d_steps}_rl_checkpoints_reinvent_lr_8e-5",

    )
```

## Evaluation

```python
import torch
from rdkit import Chem, RDLogger
from ml_conformer_generator.src.mlconfgen import MLConformerGenerator, evaluate_samples, ReinventScoreWrapper
from ml_conformer_generator.src.mlconfgen.utils import samples_to_rdkit_mol, ATOM_DECODER, prepare_adj_mat_seer_input, redefine_bonds, standardize_mol

RDLogger.DisableLog('rdApp.*')

if torch.cuda.is_available():
    device = torch.device("cuda:0")
elif torch.backends.mps.is_available():
    device = torch.device("mps:0")
else:
    device = torch.device("cpu")

def edm_sampler_fn(generator, task: dict,) -> list[Chem.Mol]:
    ref_context, ref_n_atoms, fixed_fragment = generator.prepare_inputs(
        reference_conformer=task["reference_conformer"],
        fixed_fragment=None,
        reference_context=None,
        n_atoms=None,
    )

    min_n_nodes = ref_n_atoms - task["variance"]
    max_n_nodes = ref_n_atoms + task["variance"]
    return generator.edm_samples(
        reference_context=ref_context,
        n_samples=task["n_samples"],
        min_n_nodes=min_n_nodes,
        max_n_nodes=max_n_nodes,
        raw_output=True,
    )


def set_mol_props(mol, name, ref_name, reinvent_score, shape_score):
    mol.SetProp("_Name", name)
    mol.SetProp("reference_mol", ref_name)
    mol.SetProp("reinvent_score", str(reinvent_score))
    mol.SetProp("shape_score", str(shape_score))
    return None


@torch.no_grad()
def evaluate(
             bl_generator,
             ft_generator,
             score_fn: callable,
             task: dict,
             agent_writer,
             baseline_writer,
             ) -> dict[str, float]:

    ref_name = task["reference_conformer"].GetProp("_Name")

    x, h, node_mask, edge_mask = edm_sampler_fn(bl_generator, task)

    _x, _h, edm_adapter_log_probs, edm_aux = ft_generator.edm_adapter(
        x=x,
        h=h,
        node_mask=node_mask,
        edge_mask=edge_mask,
        sample=False,
    )

    ft_samples = samples_to_rdkit_mol(
        positions=_x, one_hot=_h, node_mask=node_mask, atom_decoder=ATOM_DECODER
    )

    baseline_samples = samples_to_rdkit_mol(
        positions=x, one_hot=h, node_mask=node_mask, atom_decoder=ATOM_DECODER
    )

    (
        el_batch,
        dm_batch,
        b_adj_mat_batch,
        canonicalised_samples,
    ) = prepare_adj_mat_seer_input(
        mols=ft_samples,
        dimension=ft_generator.dimension,
        device=ft_generator.device,
    )

    (
        bl_el_batch,
        bl_dm_batch,
        bl_b_adj_mat_batch,
        bl_canonicalised_samples,
    ) = prepare_adj_mat_seer_input(
        mols=baseline_samples,
        dimension=bl_generator.dimension,
        device=bl_generator.device,
    )

    agent_adj_mat_batch = ft_generator.adj_mat_seer(
        elements=el_batch,
        dist_mat=dm_batch,
        adj_mat=b_adj_mat_batch,
    )

    baseline_adj_mat_batch = bl_generator.adj_mat_seer(
        elements=bl_el_batch,
        dist_mat=bl_dm_batch,
        adj_mat=bl_b_adj_mat_batch,
    )

    agent_mols = []
    baseline_mols = []

    # agent_valid_flags = []
    # baseline_valid_flags = []

    agent_adj_mat_batch.to("cpu")
    baseline_adj_mat_batch.to("cpu")

    for i, base_mol in enumerate(canonicalised_samples):
        agent_adj_mat = agent_adj_mat_batch[i]
        baseline_adj_mat = baseline_adj_mat_batch[i]
        _baseline_mol = bl_canonicalised_samples[i]

        agent_mol = redefine_bonds(mol=base_mol, adj_mat=agent_adj_mat)
        baseline_mol = redefine_bonds(mol=_baseline_mol, adj_mat=baseline_adj_mat)

        # Standardize mols
        std_agent_mol = standardize_mol(
                mol=agent_mol, optimize_geometry=True, ifm_mode=False
            )

        if std_agent_mol:
          agent_mols.append(std_agent_mol)


        std_baseline_mol = standardize_mol(
                mol=baseline_mol, optimize_geometry=True, ifm_mode=False
            )

        if std_baseline_mol:
          baseline_mols.append(std_baseline_mol)

    ref_mol = task['reference_conformer']


    reinvent_agent_scores = score_fn(agent_mols)

    _, aligned_agent_samples = evaluate_samples(ref_mol, agent_mols)
    shape_agent_scores = [x['shape_tanimoto'] for x in aligned_agent_samples]

    reinvent_baseline_scores = score_fn(baseline_mols)

    _, aligned_bl_samples = evaluate_samples(ref_mol, baseline_mols)
    shape_baseline_scores = [x['shape_tanimoto'] for x in aligned_bl_samples]

    a_valid = 0
    b_valid = 0

    for i, a_mol in enumerate(agent_mols):
      name = f"AS_{i+1}"
      set_mol_props(a_mol, name, ref_name, reinvent_agent_scores[i], shape_agent_scores[i])
      agent_writer.write(a_mol)
      a_valid += 1

    for i, b_mol in enumerate(baseline_mols):
      name = f"BS_{i+1}"
      set_mol_props(b_mol, name, ref_name, reinvent_baseline_scores[i], shape_baseline_scores[i])
      baseline_writer.write(b_mol)
      b_valid += 1


    agent_scores_t = torch.tensor(reinvent_agent_scores, dtype=torch.float32)
    baseline_scores_t = torch.tensor(reinvent_baseline_scores, dtype=torch.float32)

    shape_a_scores_t = torch.tensor(shape_agent_scores, dtype=torch.float32)
    shape_bl_scores_t = torch.tensor(shape_baseline_scores, dtype=torch.float32)

    f_agent_score = agent_scores_t.mean().item()
    f_baseline_score = baseline_scores_t.mean().item()

    baseline_imporv = f_agent_score - f_baseline_score

    agent_valid = a_valid / task["n_samples"]
    baseline_valid = b_valid / task["n_samples"]

    return {
        "eval_agent_scores_mean": agent_scores_t.mean().item(),
        "eval_baseline_scores_mean": baseline_scores_t.mean().item(),
        "eval_agent_shape_scores_mean": shape_a_scores_t.mean().item(),
        "eval_baseline_shape_scores_mean": shape_bl_scores_t.mean().item(),
        "eval_agent_valid_rate": agent_valid,
        "eval_baseline_valid_rate": baseline_valid,
        "eval_improve_mean": baseline_imporv,
    }

scoring_config = """
[scoring]
type = "geometric_mean"

[[scoring.component]]
[scoring.component.custom_alerts]

[[scoring.component.custom_alerts.endpoint]]
name = "Alerts"

params.smarts = [
    "[*;r8]",
    "[*;r9]",
    "[*;r10]",
    "[*;r11]",
    "[*;r12]",
    "[*;r13]",
    "[*;r14]",
    "[*;r15]",
    "[*;r16]",
    "[*;r17]",
    "[#8][#8]",
    "[#6;+]",
    "[#16][#16]",
    "[#7;!n][S;!$(S(=O)=O)]",
    "[#7;!n][#7;!n]",
    "C#C",
    "C(=[O,S])[O,S]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#16;!s]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#7;!n]",
    "[#7;!n][C;!$(C(=[O,N])[N,O])][#8;!o]",
    "[#8;!o][C;!$(C(=[O,N])[N,O])][#16;!s]",
    "[#8;!o][C;!$(C(=[O,N])[N,O])][#8;!o]",
    "[#16;!s][C;!$(C(=[O,N])[N,O])][#16;!s]"
]

[[scoring.component]]
[scoring.component.QED]

[[scoring.component.QED.endpoint]]
name = "QED"
weight = 0.6


[[scoring.component]]
[scoring.component.NumAtomStereoCenters]

[[scoring.component.NumAtomStereoCenters.endpoint]]
name = "Stereo"
weight = 0.4

transform.type = "left_step"
transform.low = 0
"""

toml_config_filename = "reinvent_scoring.toml"

with open(toml_config_filename, "w") as tf:
    tf.write(scoring_config)



target_mol = Chem.SDMolSupplier('./rl_fine_tuning_refs.sdf')[0]

ref_name = target_mol.GetProp("_Name")

scoring_fn = ReinventScoreWrapper(toml_config_filename)



N_ITERS = 10
N_SAMPES = 100

DIFFUSION_STEPS = 100

FT_CHECKPOINT = f"./{DIFFUSION_STEPS}_steps_checkpoints/best_checkpoint.pt"

FT_SAMPLES_WRITER = Chem.SDWriter(f"{ref_name}_fine_tuned_dsteps_{DIFFUSION_STEPS}_best.sdf")
BL_SAMPLES_WRITER = Chem.SDWriter(f"{ref_name}_baseline_dsteps_{DIFFUSION_STEPS}_best.sdf")

TASK = {
    "reference_conformer": target_mol,
    "variance": 1,
    "n_samples": 100,
}

bl_generator = MLConformerGenerator(
                                    edm_weights="./edm_moi_chembl_15_39.pt",
                                    adj_mat_seer_weights="./adj_mat_seer_chembl_15_39.pt",
                                    device=device,
                                    diffusion_steps=DIFFUSION_STEPS,
                                    )

ft_generator = MLConformerGenerator(
                                    edm_weights="./edm_moi_chembl_15_39.pt",
                                    adj_mat_seer_weights="./adj_mat_seer_chembl_15_39.pt",
                                    finetune_checkpoint = FT_CHECKPOINT,
                                    device=device,
                                    diffusion_steps=DIFFUSION_STEPS,
                                    )


for i in range(N_ITERS):
  print(f"Performing {i+1} generation out of {N_ITERS}")
  results = evaluate(
             bl_generator,
             ft_generator,
             scoring_fn,
             TASK,
             FT_SAMPLES_WRITER,
             BL_SAMPLES_WRITER
             )

  print(results)



```
