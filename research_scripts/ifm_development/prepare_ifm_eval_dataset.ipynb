```python
from rdkit import Chem
from tqdm import tqdm
from src.mlconfgen.utils.mol_split import split_molecule_size_constrained

reader = Chem.SDMolSupplier('./data/1000_ccdc_validation_set.sdf')

min_size = 6
max_size = 15

counter = {}
selected = []
for mol in tqdm(reader):

    frags = split_molecule_size_constrained(
                                                    mol,
                                                    min_size=min_size,
                                                    max_size=max_size,
                                                    max_iter=200,
                                                )
    n_frags = len(frags)
    if n_frags in counter.keys():
        counter[n_frags] += 1
    else:
        counter[n_frags] = 1

    if n_frags >= 3:
        selected.append(mol)

with Chem.SDWriter('data/ifm_chain_merge_test_6_15.sdf') as w:
  for m in selected:
    w.write(m)
        

print(len(selected))
    
```
